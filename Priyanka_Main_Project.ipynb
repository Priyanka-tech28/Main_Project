```python
print("Outlier Count:")

for col in numeric_columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - (1.5 * IQR)
    upper = Q3 + (1.5 * IQR)

    outliers = ((df[col] < lower) |
                (df[col] > upper)).sum()

    print(f'{col} outlier count: {outliers}')
```

```python
#Select the numerical columns
#Let's start with the columns where you previously identified outliers.
columns = [
    'Unit_Price_INR',
    'Sales_INR',
    'Discount_Amount_INR',
    'Net_Sales_INR',
    'Cost_INR',
    'Profit_INR',
    'Customer_Satisfaction'
]

df[columns].describe().T
```

```python
#Calculate the IQR limits
for col in columns:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    print(f"\nColumn: {col}")
    print("Lower Limit:", lower_limit)
    print("Upper Limit:", upper_limit)
```

```python
#Separate outliers from normal values
for col in columns:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = df[
        (df[col] < lower_limit) |
        (df[col] > upper_limit)
    ]

    print(f"\n{col}")
    print("Number of outliers:", len(outliers))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#Chart 1: Count Plot – Product Category vs No.of orders
plt.figure(figsize=(9,5))
sns.countplot(data=df, x='Category',
              order=df['Category'].value_counts().index,
              palette ='coolwarm',
              edgecolor='black'
              )
plt.title('Number of Orders by Category')
plt.xlabel('Product Category')
plt.ylabel('Number of Orders')
plt.grid(axis='y', alpha=0.5)
plt.xticks(rotation=45)
plt.show()
```

**Interpretation – Number of Orders by Product Category**
- A count plot is used to compare the number of orders across different product categories.
- The X-axis represents the product categories, and the Y-axis represents the number of orders in each category.
- Each bar shows how many records or orders belong to a particular category.
- The categories are arranged in descending order, so the category with the highest number of orders appears first.
- This chart helps identify the most popular and least popular product categories based on order count.
- The business can use this information to plan inventory, improve product availability, and focus marketing strategies on categories with lower order counts.

```python
#Chart 2:Histogram-Sales Distribution (Up to ₹50,000)
plt.figure(figsize=(8,5))

sns.histplot(df['Sales_INR'], bins=30, kde=True)

plt.xlim(0, 50000)
plt.title('Sales Distribution (Up to ₹50,000)')
plt.xlabel('Sales (INR)')
plt.ylabel('Frequency')
plt.grid(axis='y', alpha=0.3)

plt.show()
```

**Interpretation – Sales Distribution**
- A histogram is used to understand the distribution and frequency of sales values in the dataset.
- The X-axis represents the sales amount in INR, and the Y-axis represents the frequency (number of transactions).
- The bins=30 parameter divides the sales values into 30 intervals to show how frequently sales fall within each range.
- The kde=True parameter adds a smooth density curve to help understand the overall shape of the sales distribution.
- The plt.xlim(0, 50000) parameter displays sales values only from ₹0 to ₹50,000, making it easier to examine the distribution within this range.
- This chart helps identify the most common sales ranges and understand whether sales values are concentrated in lower or higher ranges.
- The analysis can help the business understand its sales patterns and identify transactions with unusually high sales values.

```python
#Chart 3:Bar Chart – Top 10 Subcategories Total Sales
top_subcategories = df.groupby('Subcategory')['Sales_INR'].sum().nlargest(10).reset_index()
top_subcategories
```

```python
plt.figure(figsize=(9,6))
sns.barplot (
    data=top_subcategories,
    x='Sales_INR',
    y='Subcategory',
    palette='Set2'
)

plt.title('Top 10 Subcategories by Total Sales')
plt.xlabel('Total Sales (INR)')
plt.ylabel('Subcategory')
plt.grid(axis='x', alpha=0.3)
plt.show()
```

**Interpretation – Top 10 Subcategories by Total Sales**
- A bar plot is used to identify the top 10 product subcategories that generate the highest total sales.
- The X-axis represents the total sales amount in INR, and the Y-axis represents the product subcategories.
- Total sales are calculated by grouping the data based on Subcategory and adding the Sales_INR values for each subcategory.
- The nlargest(10) function selects the 10 subcategories with the highest total sales.
- The chart helps identify the best-performing subcategories and compare their contribution to overall sales.
- This analysis helps the business focus on high-performing products, improve inventory planning, and develop marketing strategies for subcategories with lower sales.

```python
#Chart 4:Bar plot- Total Sales by Region
plt.figure(figsize=(8,5))
sns.barplot (data=df,
            x='Category',
            y='Sales_INR',
            hue= 'Region',
            estimator='sum',
            palette= 'coolwarm',
            errorbar= 'ci')

plt.title('Total Sales by Category')
plt.xlabel('Category')
plt.ylabel('Total Sales (INR)')
plt.grid(axis='y',color='k',linestyle='--')
plt.tight_layout()
plt.show()
```

**Interpretation – Total Sales by Category**

- A bar plot is used to compare the total sales generated by different product categories.
- Category is represented on the X-axis and Total Sales (INR) is represented on the Y-axis.
- The total sales are calculated by grouping the data based on product categories and summing the sales values for each category.
- The chart helps identify which product category contributes the highest and lowest sales to the overall revenue.
- This analysis helps the business understand category-wise sales performance and make better decisions regarding product promotion, inventory management and sales strategies.

```python
#Chart 5:Barplot - Profit by Product category
plt.figure(figsize=(8,5))

sns.barplot (data=df, x='Category',
            y='Profit_INR', estimator='sum',
             palette='coolwarm',
             hue='Region',
            errorbar='ci')

plt.title('Profit by Product Category')
plt.xlabel('Category')
plt.ylabel('Total Profit')
plt.grid(axis='y',color='k',linestyle='--')
plt.tight_layout()
plt.show()
```

**Interpretation – Profit by Product Category and Region**
- A grouped bar plot is used to compare the total profit generated by different product categories across different regions.
- The X-axis represents the product categories, and the Y-axis represents the total profit in INR.
- Different colours represent different regions, making it easier to compare regional profit performance within each category.
- The estimator='sum' parameter calculates the total profit for each category and region instead of displaying the average profit.
- The chart helps identify which product categories and regions contribute more or less to the reported profit.
- This analysis can help the business understand regional performance, identify profitable categories, and plan suitable sales and marketing strategies.

```python
#Chart 6:Scatter Plot – Discount vs Sales
plt.figure(figsize=(8,5))
sns.scatterplot(
    data=df,
    x='Discount_Amount_INR',
    y='Sales_INR',
    palette='viridis',alpha=0.6,s=80
)

plt.title('Discount vs Sales')
plt.xlabel('Discount Amount (INR)')
plt.ylabel('Sales (INR)')
plt.legend(bbox_to_anchor=(1, 1))
plt.grid(axis='y',color='k',linestyle='--')
plt.tight_layout()
plt.show()
```

**Interpretation – Discount vs Sales**
- A scatter plot is used to analyse the relationship between discount amounts and sales values.
- The X-axis represents the discount amount in INR, and the Y-axis represents the sales amount in INR.
- Each dot represents an individual transaction in the dataset.
- The chart helps identify whether higher discount amounts are associated with higher or lower sales values.
- The distribution of points can help identify patterns, relationships, and unusual transactions between discounts and sales.
- This analysis helps the business understand discounting patterns and evaluate how discounts are associated with sales performance.

```python
#chart 7:Heatmap – Correlation Analysis
plt.figure(figsize=(11,7))

corr = df.select_dtypes(include='number').corr()

sns.heatmap(
    corr,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Correlation Heatmap')
plt.show()
```

**Interpretation – Correlation Heatmap**
- A correlation heatmap is used to identify the strength and direction of relationships between numerical columns in the dataset.
- It helps understand how different numerical variables, such as Quantity, Sales_INR, Discount_Amount_INR, Net_Sales_INR, Cost_INR, and Profit_INR, are related to one another.
- The correlation values range from -1 to +1:
  - +1: Perfect positive correlation – both variables move in the same direction.
  - -1: Perfect negative correlation – one variable increases while the other decreases.
  - 0: No linear correlation between the variables.
- The colour intensity helps identify strong and weak relationships between numerical variables.
- This chart helps identify important relationships, understand sales patterns, and select relevant variables for further analysis.

```python
#Chart 8:Quantity Distribution by Category
plt.figure(figsize=(10,5))
sns.boxplot(
    data=df,
    x='Category',
    y='Quantity',
    palette='Set2'
)

plt.title('Quantity Distribution by Category')
plt.xlabel('Category')
plt.ylabel('Quantity Sold')
plt.xticks(rotation=45)
plt.grid(axis='y',color='k',linestyle='--')
plt.tight_layout()
plt.show()
```

**Interpretation – Quantity Distribution by Category**
- A box plot is used to understand and compare the distribution of quantities sold across different product categories.

```python
#chart 9:Pie-Customer Satisfaction Rating Distribution
rating_counts = df['Customer_Satisfaction'].value_counts().sort_index()

plt.figure(figsize=(8,8))

plt.pie(
    rating_counts,
    labels=rating_counts.index,
    autopct='%1.1f%%',
    startangle=90,
    colors=['skyblue', 'orange', 'lightgreen', 'pink', 'yellow'],
    explode=[0.05, 0, 0, 0, 0],
    shadow=True
)

plt.title('Customer Satisfaction Rating Distribution')
plt.show()
```

**Interpretation – Customer Satisfaction Rating Distribution**
- A pie chart is used to visualize the percentage distribution of customer satisfaction ratings.
- Each slice represents a customer satisfaction rating from 1 to 5.
- The size of each slice indicates the percentage of customers who gave that particular rating.
- The autopct parameter displays the percentage contribution of each rating.
- The chart helps identify the most common and least common customer satisfaction ratings.
- It provides an overview of customer satisfaction levels and helps the business understand how customers feel about its products or services.
- The business can use these insights to improve product quality, customer experience, and service strategies.

```python
#chart 10:Line plot-Monthly Sales vs Net Sales Trend
df['Order_Date'] = pd.to_datetime(
    df['Order_Date'], dayfirst=True, errors='coerce'
)

monthly_data = df.groupby(
    df['Order_Date'].dt.to_period('M')
)[['Sales_INR', 'Net_Sales_INR']].sum()

monthly_data.index = monthly_data.index.astype(str)

monthly_data.plot(figsize=(12,6), marker='o')

plt.title('Monthly Sales vs Net Sales Trend')
plt.xlabel('Month')
plt.ylabel('Amount (INR)')
plt.xticks(rotation=45)
plt.legend(['Total Sales', 'Net Sales'])
plt.grid(alpha=0.3)
plt.show()
```

**Interpretation – Monthly Sales vs Net Sales Trend**
- A line chart is used to analyse the monthly trend of total sales and net sales over time.
- The X-axis represents the months, and the Y-axis represents the total sales amount in INR.
- The chart compares two measures: Total Sales and Net Sales for each month.
- Total Sales represents the sales amount before discounts, whereas Net Sales represents the sales amount after deducting discounts.
- The difference between the two lines helps understand how discounts affect the monthly sales value.
- The chart helps identify monthly sales growth, declining trends, seasonal patterns, and fluctuations in revenue.
- This analysis supports business decisions related to discount strategies, sales planning, and revenue improvement.
